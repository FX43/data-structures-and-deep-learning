### 1.3 队列
特点：先进先出(FIFO)

1.3.1 定义

```python
class Queue:
    def __init__(self):
        self.items=[]
        self.front=0

    def enqueue(self,item):
        self.items.append(item)

    def dequeue(self):
        return self.items.pop(0)

    def isEmpty(self):
        return not self.items

    def peek(self):
        return self.items[self.front]

    def __len__(self):
        return len(self.items)
```

```python
def transfer(name_list,num):
    q=Queue()
    for i in name_list:
        q.enqueue(i)
    while len(q) > 1:
        for i in range(num):
            q.enqueue(q.dequeue())
        q.dequeue()

    return q.items

name_list =[str(x) for x in range(10)]
print(transfer(name_list,5))
```

```python
import random

class newTask:
    def __init__(self, start_time):
        self.page = random.randint(1,21)
        self.start_time = start_time


class printer:
    def __init__(self,speed_print):
        self.page = 0
        self.speed = speed_print
        self.time = 0

    def convert(self,new_page):
        self.time = new_page / self.speed

    def work(self):
        if self.time > 0:
            self.time -= 1

    def isBusy(self):
        return  self.time

def isCreate():
    return random.randint(1,180)==1

def simulation(sim_time):
    q=Queue()
    a_printer = printer(1/6)
    wait_list = []

    for time in range(sim_time):
        if isCreate():
            new_task = newTask(time)
            q.enqueue(new_task)
        if not a_printer.isBusy() and q.items:
            dequeue = q.dequeue()
            wait_list.append(time - dequeue.start_time)
            a_printer.convert(dequeue.page)

        a_printer.work()
    aver_time = sum(wait_list) / len(wait_list)
    return "平均等待时间为 %.2f,目前还有 %d 任务没有打印"%(aver_time,len(q))

for sim_time in range(10):
    print(simulation(3600))
```
