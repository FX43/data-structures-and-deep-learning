并查集数据结构

基本功能：

1.make_set:让每个元素自成一个子集合

2.find:找到当前节点的根节点

3.union:合并两个集合

```python
class UnionFound:
    def __init__(self,length):
        self.a = [i for i in range(length)]#索引代表某个元素，索引上的值代表这个元素父亲的索引
        self.rank = [0]*length #即可以按照值的大小进行合并，也可以按照节点个数大小合并

    def Findparent(self,x):#路径压缩，通过递归的方法使得索引为x的位置直接指向根节点
        if x != self.a[x]:
            self.a[x] = self.Findparent(self.a[x])#
        return self.a[x]

    def Union(self,x,y):#按秩合并
        x_root = self.Findparent(x)
        y_root = self.Findparent(y)
        if x_root == y_root:
            return
        if self.rank[x_root] < self.rank[y_root]:
            self.a[x_root] = y_root
        elif self.rank[x_root] < self.rank[y_root]:
            self.a[y_root] = x_root
        else:
            self.a[y_root] = x_root
            self.rank[x_root] += 1

a = UnionFound()#这是实例化，但存在问题，因为没有给定length的值
b = UnionFound#这不是实例化，只是将UnionFound这个类赋值给b
```
