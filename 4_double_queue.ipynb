1.4 双端队列

1.4.1 实现

```python
class Deque:
    def __init__(self):
        self.queue = []
    def addFront(self, value):
        self.queue.append(value)
    def addRear(self, value):
        self.queue.insert(0, value)
    def removeRear(self):
        return self.queue.pop(0)
    def removeFront(self):
        return self.queue.pop()
    def isEmpty(self):
        return self.queue == []
    def __len__(self):
        return len(self.queue)
```

```python
def huiwen(string):
    flag = True
    q = Deque()
    #(char for char in string)生成表达式
    for char in string:
        q.addFront(char)
    while len(q) > 1 and flag:
        if q.removeRear() != q.removeFront():
            flag = False

    return flag

string = "ws"
huiwen(string)
```

        (char for char in string) 生成器表达式
* 使用圆括号 ()
* 返回一个生成器对象
* 惰性求值：不会立即计算所有元素
* 内存高效：一次只生成一个元素，适合处理大数据集
* 只能迭代一次：使用后生成器会耗尽

```python
string = "abc"
gen_expr = (char for char in string)
print(gen_expr)  # 输出: <generator object <genexpr> at 0x...>

# 需要转换为列表或使用循环来获取值,都只能使用一次
#print(list(gen_expr))  # 输出: ['a', 'b', 'c']
for i in gen_expr:
    print(i)

for i in gen_expr:
    print(i)
```

        [char for char in string] 列表表达式
* 使用方括号 []
* 返回一个列表对象
* 立即求值：一次性计算所有元素
* 内存消耗大：所有元素都存储在内存中
* 可多次使用：列表可以重复访问

```python
string = "abc"
list_comp = [char for char in string]
print(list_comp)  # 输出: ['a', 'b', 'c']

# 可以直接使用
print(list_comp[0])  # 输出: 'a'
```
