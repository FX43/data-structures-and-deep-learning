## 200-1 岛屿数量


### 解题思路


### 方法一：DFS


```python
class Solution(object):
    def numIslands(self, grid):
        """
        :type grid: List[List[str]]
        :rtype: int
        """
        m = len(grid)
        n = len(grid[0])
        num = 0

        def dfs(i,j):
            if i < 0 or i > m-1 or j < 0 or j > n-1:
                return
            if grid[i][j] == "0" or grid[i][j] == "2":
                return
            grid[i][j] = "2"
            dfs(i-1,j)
            dfs(i+1,j)
            dfs(i,j+1)
            dfs(i,j-1)

        for i in range(m):
            for j in range(n):
                if grid[i][j] == "1":
                    #下面不能写grid[i][j] = "0"，否则进入递归后直接退出
                    dfs(i,j)
                    num += 1

        return num
```

### 方法二：BFS  对于坐标而言，用元组存储更加方便

deque（双端队列数据结构）的用法：

基本操作：

        quene = deque([i,j]): 初始化
        d.appendleft() :从左端添加元素
        d.append():从右端添加元素
        d.pop(): 从右端弹出元素
        d.popleft(): 从左端弹出元素

优势：
    双端队列两端入队与出队的时间复杂度都为O(1)


```python
from collections import deque

class Solution(object):
    def numIslands(self, grid):
        if not grid:
            return 0

        m = len(grid)
        n = len(grid[0])

        def bfs(i, j):
            queue = deque([(i, j)])
            grid[i][j] = "0"  # 标记为已访问

            while queue:
                x, y = queue.popleft()#这种情况下x与y都是数字类型，相当于提前解包，如果左端只有一个变量的话那么它是一个列表或者元组
                # 四个方向
                for dx, dy in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
                    nx, ny = x + dx, y + dy
                    if 0 <= nx < m and 0 <= ny < n and grid[nx][ny] == "1":#坐标要合理,同时是没有被探索过的坐标
                        grid[nx][ny] = "0"
                        queue.append((nx, ny))

        num = 0
        for i in range(m):
            for j in range(n):
                if grid[i][j] == "1":
                    bfs(i, j)
                    num += 1

        return num
```

## 994-2 腐烂的橘子


### 腐烂橘子问题


```python
class Solution:
    def orangesRotting(self, grid: List[List[int]]) -> int:
        m,n = len(grid),len(grid[0])
        queue = deque()
        num,minute = 0,0 #用于记录好橘子以及腐烂用时
        for i in range(m):
            for j in range(n):
                if grid[i][j] == 2:
                    queue.append([i,j])
                if grid[i][j] == 1:
                    num += 1

        if not num:
            return 0

        while queue:
            if not num:
                return minute
            length = len(queue)
            for _ in range(length):
                x,y = queue.popleft()
                for dx,dy in [(0,1),(0,-1),(1,0),(-1,0)]:
                    nx,ny = x+dx,y+dy
                    if 0 <= nx < m and 0<= ny < n and grid[nx][ny] == 1:
                        grid[nx][ny] = 2
                        queue.append([nx,ny])
                        num -= 1
            minute += 1
        return -1
```

## 207-3 课程表


### 课程表问题（拓扑排序问题）

采用入度出度算法

当从队列中弹出某个元素的时候，只需对这个元素的子节点的入度减一即可，使用邻接表数据结构保存每个节点的子节点都有谁，在进行入度减一操作时直接在邻接表中进行而不需要遍历所有节点

对于邻接表数据结构：索引代表某个节点，索引上的列表中是这个节点所有的子节点


```python
class Solution(object):
    def canFinish(self, numCourses, prerequisites):
        """
        :type numCourses: int
        :type prerequisites: List[List[int]]
        :rtype: bool
        """
        indegree = [0] * numCourses
        adject = [[] for _ in range(numCourses)]
        num = 0

        for child,par in prerequisites:
            adject[par].append(child)#存储这个节点的子节点，方便后续子节点的入度减一操作
            indegree[child] += 1

        queue = deque([i for i in range(numCourses) if indegree[i]==0])#构建一个双端队列，其中里面的元素都入读都为0

        while queue:
            pop = queue.popleft()
            num += 1

            for j in adject[pop]:
                indegree[j] -= 1
                if indegree[j] == 0:
                    queue.append(j)

        return True if num == numCourses else False
```

    list1 = [[] * n] : 错误
    list1 = [[] for _ in range(n)] :正确

## 208-4 实现 Trie (前缀树)


### 生成前缀树

生成前缀树的思想核心是trienode节点，这个节点本身代表一个字母，哈希表里面存储的是以这个字母为前缀的其他字母，通过哈希表可以直接访问到这些字母节点


```python
class Trienode:
    def __init__(self):
        self.hashtable = {}
        self.flag = False

class Trie:

    def __init__(self):
        self.root = Trienode()

    def insert(self, word: str) -> None:
        node = self.root
        for char in word:
            if char not in node.hashtable:
                node.hashtable[char] = Trienode()
            node = node.hashtable[char]
        node.flag = True

    def search(self, word: str) -> bool:
        node = self.root
        for char in word:
            if char not in node.hashtable:
                return False
            node = node.hashtable[char]
        return True if node.flag else False

    def startsWith(self, prefix: str) -> bool:
        node = self.root
        for char in prefix:
            if char not in node.hashtable:
                return False
            node = node.hashtable[char]
        return True
```
