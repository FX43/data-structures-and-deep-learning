## 136-5 只出现一次的数字

### 解题思路

对数组全部元素做异或：利用 a ^ a = 0、a ^ 0 = a，成对元素相互抵消，剩下的就是只出现一次的数字。

### Python 中的异或 `^`

Python 中的 `^` 表示**按位异或（bitwise XOR）**：

```python
a ^ b
```

计算时会先把两个整数写成二进制，然后**逐位进行异或**：

$$
0\oplus0=0,\quad
0\oplus1=1,\quad
1\oplus0=1,\quad
1\oplus1=0
$$

也就是：

$$
\boxed{\text{相同为 0，不同为 1}}
$$

### 常用性质

$$
\boxed{a\oplus a=0}
$$

$$
\boxed{a\oplus 0=a}
$$

$$
\boxed{a\oplus b=b\oplus a}
$$

> **一句话记忆：** Python 的 `^` 是对整数的二进制位逐位异或，**相同为 0，不同为 1**。

````python
class Solution:
    def singleNumber(self, nums: List[int]) -> int:
        return reduce(xor, nums)#用 func 不断把序列中的元素合并成一个值。
````

## 75-5 颜色分类


### 荷兰国旗方法（简化版的三路快速排序）


荷兰国旗三指针:三个指针分别为lt,i,rt,其中 **[0,lt]元素全为0,[lt+1,i)元素全为1;[i,gt)为探索区域;[gt,right]内部元素全为2** .其中[]代表包含对应元素,()代表不包含对应元素

````python
class Solution(object):
    def sortColors(self, nums):
        """
        :type nums: List[int]8
        :rtype: None Do not return anything, modify nums in-place instead.
        """
        left = 0
        right = len(nums) - 1

        lt = left - 1
        i = 0
        gt = right + 1

        while i < gt: #未探索区间为[i,gt),因此当二者相等的时候说明所有元素判断完成,结束循环
            if nums[i] < 1:
                lt += 1
                nums[i], nums[lt] = nums[lt],nums[i]
                i += 1
            elif nums[i] == 1:
                i += 1
            else:
                gt -= 1
                nums[i],nums[gt] = nums[gt],nums[i]
````

## 31-9 下一个排列


### 解题思路

思路：

    1.从右向左，找第一个小于右侧相邻数字的数 x
    2.找 x 右边最小的大于 x 的数 y(也就是从右向左找到的第一个数值)，交换 x 和 y
    3.反转 y (不包括y)右边的数，把右边的数变成最小的排列



````python
class Solution:
    def nextPermutation(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        n = len(nums)
        i = n - 2
        while i >= 0 and nums[i] >= nums[i+1]:
            i -= 1

        if i >= 0:
            j = n -1
            while nums[j] <= nums[i]:#因为x的右侧是递减的，从右往左找的过程就是递增的，找到的第一个大于nums[i]的值就是最小的大值
                j -= 1
            nums[j],nums[i] = nums[i],nums[j]

        left,right = i+1,n-1
        while left < right:
            nums[left],nums[right] = nums[right],nums[left]
            left += 1
            right -= 1
````

## 287-7 寻找重复数


### 寻找重复数字

这题的关键在于给出一个包含n+1个数字的数组，且数组的取值位于[1,n],可以这么看待问题，把这n+1个数对应的索引看成是n+1个节点，标号从0到n，构建从i到nums[i]的映射，nums[i]的值就代表了i这个节点指向哪个节点，因为每个节点的出度为1，(索引不会重复),就是说每个节点只能指向某一个节点，不能指向多个节点，但是一个节点可以被多个节点指向，对于数组中的重复数而言，他们就是这种情况，因此重复数节点构成了环的入口。采用环形链表II的解题思路解决这道题.




````python
class Solution:
    def findDuplicate(self, nums) -> int:
        n = len(nums)
        fast = slow = 0#从节点0开始出发

        while True:
            fast = nums[nums[fast]]#映射两次代表走两步
            slow = nums[slow]

            if fast == slow:
                slow = 0
                while fast != slow:
                    fast = nums[fast]
                    slow = nums[slow]
                return fast
````
