**53. 最大子数组和**

普通数组 / 连续子数组 · Hot100 · 中等

[官网题目](https://leetcode.cn/problems/maximum-subarray/) · [返回本文目录](#本文目录) · [总目录](README.md)

笔记情况：已有详细笔记。解法标签：动态规划 / 普通数组。

表格摘要：Kadane算法的核心思想是：动态规划 遍历数组，对于每个位置，决定是继续扩展当前子数组，还是从当前位置开始新的子数组 通过比较"当前元素值"与"当前元素值+之前子数组和"来做出决策


**_#1.最大连续子数组之和_**

Kadane算法的核心思想是：动态规划

遍历数组，对于每个位置，决定是继续扩展当前子数组，还是从当前位置开始新的子数组

通过比较"当前元素值"与"当前元素值+之前子数组和"来做出决策

```python
class Solution(object):
    def maxSubArray(self, nums):

        curr_sum = max_sum =nums[0]

        for i in nums[1:]:
            curr_sum = max(i,curr_sum+i)#如果i更大，意味着前面的数组贡献为负，如果curr_sum+i，意味着前面的数组贡献为正。
            max_sum = max(max_sum,curr_sum)
        return max_sum
```

## lc-56

**56. 合并区间**

普通数组 / 区间合并 · Hot100 · 中等

[官网题目](https://leetcode.cn/problems/merge-intervals/) · [返回本文目录](#本文目录) · [总目录](README.md)

笔记情况：已有详细笔记。解法标签：数组 / 排序。

表格摘要：当需要合并的时候，不用考虑将元素弹出再将新的元素加入，只需将数组最后一个元素的值进行更改即可


**#5.合并区间**

当需要合并的时候，不用考虑将元素弹出再将新的元素加入，只需将数组最后一个元素的值进行更改即可

```python
class Solution:
    def merge(self, intervals: List[List[int]]) -> List[List[int]]:

    intervals.sort(key = lambda x:x[0])
    merged = []

    for interval in intervals:
        if not merged or merged[-1][1] < interval[0]:
            merged.append(interval)
        else:
            max_num = max(interval[1],merged[-1][1])
            merged[-1][1] = max_num
    return merged
```

## lc-189

**189. 轮转数组**

普通数组 / 原地变换 · Hot100 · 中等

[官网题目](https://leetcode.cn/problems/rotate-array/) · [返回本文目录](#本文目录) · [总目录](README.md)

笔记情况：已有详细笔记。解法标签：数组 / 数学 / 双指针。

表格摘要：1. nums[:] = nums[length-k:]+nums[:length-k] 2. nums = nums[length-k:]+nums[:length-k] 这是两个完全不同的操作，操作1对nums指向的…


**_#4.轮转数组_**

1.         nums[:] = nums[length-k:]+nums[:length-k]
2.         nums = nums[length-k:]+nums[:length-k]

这是两个完全不同的操作，操作1对nums指向的数组进行修改，而操作二只是将nums指针指向了新的数组，对原来的数组没有修改。使用切片操作会创建新的列表，空间复杂度为O(k)。+ 操作将两个列表连接，创建第三个长度为 n 的新列表

使用操作1的一个好处是不用返回值，就如同c语言中使用指针作为参数。

```python
class Solution(object):
    def rotate(self, nums, k):
        """
        :type nums: List[int]
        :type k: int
        :rtype: None Do not return anything, modify nums in-place instead.
        """
        length = len(nums)
        k = k%length
        nums[:] = nums[length-k:]+nums[:length-k]
        return nums
```

方法二：三次反转法

```python
class Solution:
    def rotate(self, nums: List[int], k: int) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        def reverse(arr, start, end):
            while start < end:
                arr[start], arr[end] = arr[end], arr[start]
                start += 1
                end -= 1

        n = len(nums)
        k %= n

        # 1. 反转整个数组
        reverse(nums, 0, n - 1)
        # 2. 反转前 k 个元素
        reverse(nums, 0, k - 1)
        # 3. 反转剩余元素
        reverse(nums, k, n - 1)
```

**238. 除了自身以外数组的乘积**

普通数组 / 前后缀乘积 · Hot100 · 中等

[官网题目](https://leetcode.cn/problems/product-of-array-except-self/) · [返回本文目录](#本文目录) · [总目录](README.md)

笔记情况：已有详细笔记。解法标签：数组 / 前缀和。

表格摘要：思路：求每个数组元素的前缀与后缀乘积，然后令前后缀相乘


**_#2.除自身以外数组的乘积_**

思路：求每个数组元素的前缀与后缀乘积，然后令前后缀相乘

```python
class Solution(object):
    def productExceptSelf(self, nums):
        """
        :type nums: List[int]
        :rtype: List[int]
        """
        length = len(nums)
        res = [1]*length
        left,right = 1,1
        for i in range(length):
            res[i] = left
            left *= nums[i]
        for j in range(length-1,-1,-1):
            res[j]*=right#这里是在前缀的基础上乘上后缀的值
            right *= nums[j]

        return res
```

## lc-41

**41. 缺失的第一个正数**

普通数组 / 原地哈希 · Hot100 · 困难

[官网题目](https://leetcode.cn/problems/first-missing-positive/) · [返回本文目录](#本文目录) · [总目录](README.md)

笔记情况：已有详细笔记。解法标签：数组 / 哈希表。

表格摘要：思想： 看学号为1到n的学生是否坐在了0到n-1的位置上 对于一个长度为n的数组，它的索引范围为[0,n-1],缺失的第一个正数一定在[1,n+1]的范围内，所以对于一个学号为nums[i]的学生，他应该坐在nums[i…


**_#8.缺失的第一个正数：_**

思想：

看学号为1到n的学生是否坐在了0到n-1的位置上



对于一个长度为n的数组，它的索引范围为[0,n-1],缺失的第一个正数一定在[1,n+1]的范围内，所以对于一个学号为nums[i]的学生，他应该坐在nums[i]-1的下标处，因此当我们只需判断nums[i]的值是否与i+1相等即可，如果不相等说明i+1就是缺失的第一个正数

```python
class Solution:
    def firstMissingPositive(self, nums) -> int:
        n = len(nums)

        for i in range(n):
            while 1 <= nums[i] <= n and nums[i] != nums[nums[i] - 1]:#当学号位于【1，n】的范围内并且该nums[i]的真身没有坐在nums[i]-1的位置上时，需要进行交换，否则的话不用进入循环，交换完毕后还需要考虑当前的nums[i]的情况
                j = nums[i] - 1
                nums[i],nums[j] = nums[j],nums[i]
                #不能直接使用nums[i], nums[nums[i] - 1] = nums[nums[i] - 1], nums[i]进行交换，当执行这样的同时赋值时，nums[i] 的值在计算索引 nums[i] - 1 之前就已经改变了。
        for i in range(n):
            if i + 1 != nums[i]:
                return i + 1
        return n + 1
```
