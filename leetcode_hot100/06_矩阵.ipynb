## 73-4 矩阵置零


### 解题思路

外部循环为(1,m),内部循环范围为(n)


```python
class Solution:
    def setZeroes(self, matrix: List[List[int]]) -> None:
        m, n = len(matrix), len(matrix[0])
        first_row_has_zero = 0 in matrix[0]

        for i in range(1, m):  
            for j in range(n):
                if matrix[i][j] == 0:
                    matrix[i][0] = matrix[0][j] = 0

        for i in range(1, m):
            # 倒着遍历，避免提前把 matrix[i][0] 改成 0，误认为这一行要全部变成 0，因为如果正序遍历的话，可能导致matrix[i][0]=0,从而使程序误判认为这一行都要置零，但实际上这个0不是原生的0.
            # 倒序就是为了防止 matrix[0][0] = 0 在处理第 0 列时提前把 matrix[i][0] 改成 0，进而误伤整行。
            for j in range(n - 1, -1, -1):
                if matrix[i][0] == 0 or matrix[0][j] == 0:
                    matrix[i][j] = 0

        if first_row_has_zero:
            for j in range(n):
                matrix[0][j] = 0
```

## 54-1 螺旋矩阵


### 螺旋矩阵问题


```python
class Solution(object):
    def spiralOrder(self, matrix):
        top,bottom = 0,len(matrix)-1
        left,right = 0,len(matrix[0])-1
        res = []

        while left <= right and top <= bottom:
            for i in range(left,right+1):
                res.append(matrix[top][i])
            top += 1#当top=bottom时，top+=1会使得top>bottom，从而导致在接下来的循环无法进行，从行的角度理解，当top等于bottom的时候意味着只有最后一行没有取完了，而当top大于bottom则意味着所有的值都已经取完，因此后续的循环都没有继续下去的必要了

            for i in range(top,bottom+1):
                res.append(matrix[i][right])
            right -= 1

            if top <= bottom:
                for i in range(right,left-1,-1):
                    res.append(matrix[bottom][i])
                bottom -= 1

            if left <= right:
                for i in range(bottom,top-1,-1):
                    res.append(matrix[i][left])
                left += 1

        return res
```

## 48-2 旋转图像


### 旋转矩阵

将矩阵顺时针旋转90度，可以先将矩阵进行转置，然后再将每一行反转


```python
class Solution(object):
    def rotate(self, matrix):
        """
        :type matrix: List[List[int]]
        :rtype: None Do not return anything, modify matrix in-place instead.
        """
        row = len(matrix)
        col = len(matrix[0])

        for i in range(row):
            for j in range(i,col):
                matrix[i][j] , matrix[j][i] = matrix[j][i] , matrix[i][j]

        for i in range(row):
            matrix[i].reverse()#用于将每一行反转
```

## 240-3 搜索二维矩阵 II


### 解题思路

* 方法一：暴力枚举
* 方法二：每一行进行二分查找
* 方法三：从左下角或者右上角开始遍历


```python
class Solution:
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:
        m,n = len(matrix),len(matrix[0])
        row,col = 0,n-1
        while row < m and col >= 0:
            if matrix[row][col] > target:#满足这个条件的时候,不可能出现在这一列当中
                col -= 1
            elif matrix[row][col] < target:#满足这个条件,不可能满足在这一行当中
                row += 1
            else:
                return True
        return False
```
