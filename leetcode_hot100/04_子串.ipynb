## 560-3 和为 K 的子数组


### 和为k的子数组

这个题不能采用滑动窗口算法，因为不满足单调性，数组中存在负数，思路与二叉树中的路径之和III相似，使用哈希表与前缀求和的方式，具体参考路径之和III。


```python
class Solution(object):
    def subarraySum(self, nums, k):
        """
        :type nums: List[int]
        :type k: int
        :rtype: int
        """
        hashtable = Counter()
        hashtable[0] = 1
        curr_sum = 0
        ans = 0
        for left in nums:
            curr_sum += left
            ans += hashtable[curr_sum - k]#这个值存在，意味着存在这样的路径，不一定只有一条
            hashtable[curr_sum] += 1#存储的是前缀之和
        return ans
```

## 239-1 滑动窗口最大值


### 解题思路

算法的核心思想是维护一个单调递减的队列，当新元素的值比队列末尾的元素值要大的时候，那么末尾元素就不可能作为后续的最大值了，（但这个元素可能作为以它为结尾的最大值，但后续因为滑动窗口始终包含新旧元素，因此旧元素不可能是最大值），因此将其直接去除即可，而当新元素的值比老元素要小时，当老元素弹出后，新元素还是可能作为最大值的，因此加入队列


```python
class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        queue = deque()
        res = []
        for i,x in enumerate(nums):
            while queue and nums[queue[-1]] <= x:#注意这里是小于等于号
                queue.pop()
            queue.append(i)

        if i - queue[0] == k:#此时滑动窗口长度溢出，需要除去开头元素,因为此时区间里面有k+1个元素了,所以需要除去开头元素,一个区间[left,right]的长度为right-left+1
            queue.popleft()

        if i >= k-1:#当我的索引为k-1的时候，就意味着我至少已经见过了k个元素，滑动窗口形成，就可以输出最大值了
            res.append(nums[queue[0]])

        return res
```

## 76-7 最小覆盖子串


### 最小覆盖字串（从满足条件到不满足条件）


```python
class Solution:
    def minWindow(self, s: str, t: str) -> str:
        # 注：defaultdict 比 Counter 快
        diff = defaultdict(int)  # 窗口每种字母个数 - t 每种字母个数
        for c in t:
            diff[c] -= 1
        kinds = len(diff)  # t 中有 kinds 种不同的字母

        ans_left, ans_right = -1, len(s)
        ge_cnt = 0  # 窗口内有 ge_cnt 种字母的出现次数 >= t 中相应字母的出现次数
        left = 0

        for right, c in enumerate(s):  # 移动子串右端点
            diff[c] += 1  # 右端点字母移入子串
            if diff[c] == 0:  # 原来窗口内 c 的出现次数比 t 的少，现在一样多
                ge_cnt += 1  # 从 < 变成 >=

            while ge_cnt == kinds:  # 涵盖：所有字母的出现次数都是 >=
                if right - left < ans_right - ans_left:  # 找到更短的子串
                    ans_left, ans_right = left, right  # 记录此时的左右端点

                x = s[left]  # 左端点字母
                if diff[x] == 0:
                    # x 移出窗口之前，检查出现次数，
                    # 如果窗口内 x 的出现次数和 t 一样，
                    # 那么 x 移出窗口后，窗口内 x 的出现次数比 t 的少
                    ge_cnt -= 1  # 从 >= 变成 <
                diff[x] -= 1  # 左端点字母移出子串
                left += 1

        return "" if ans_left < 0 else s[ans_left: ans_right + 1]
```
