### 表一：`nn.Module` 的三大核心存储区对比

这是 PyTorch 模型内部管理数据的“三个抽屉”，理解它们的区别是核心考点。

| 属性名称 | 核心定义与作用 | 是否算梯度 (requires\_grad) | 优化器是否更新 | 是否存入 state\_dict | 典型代表 | 如何在代码中注册 (添加) |
| :--- | :--- | :---: | :---: | :---: | :--- | :--- |
| **`parameters`** | **模型参数**：需要通过反向传播去“学习”的权重。 | ✅ **是** (默认) | ✅ **是** | ✅ **是** | 线性层的 `weight` / `bias`、卷积核 | 1. 赋值 `nn.Parameter` 对象<br>2. 赋值 PyTorch 内置层 |
| **`buffers`** | **模型状态/缓存**：不需要算梯度，但属于模型固定状态，需长期保存。 | ❌ **否** (通常) | ❌ **否** | ✅ **是** | BatchNorm 的 `running_mean` (均值) 和方差 | 必须显式调用 `self.register_buffer('名字', tensor)` |
| **`modules`** | **子模块/子网络**：模型中嵌套的其他模型，像搭积木一样。 | - | - | - | `nn.Linear`、`nn.Conv2d`、`nn.Sequential` | 直接将实例化的 `nn.Module` 赋值给 self 的属性 |

### 表二：`nn.Module` 常用成员函数与变量速查

这是日常写训练代码（Training Loop）时最常调用的方法。

| 分类 | 名称 / 方法 | 核心作用与一句话笔记 | 使用场景 / 注意事项 |
| :--- | :--- | :--- | :--- |
| **状态变量** | `training` | 这是一个布尔值（True/False），**标识当前处于训练还是推理模式**。 | 强烈影响 `Dropout` 和 `BatchNorm` 层的行为。 |
| **核心计算** | `forward(x)` | **前向传播逻辑**：定义数据流经网络的计算过程（画计算图）。 | **必须重写**。永远不要直接调 `model.forward(x)`，而是直接调用模型 `model(x)`。 |
| **模式切换** | `train()` | 将模型设置为**训练模式** (`training=True`)。 | 开始训练循环前调用。 |
| **模式切换** | `eval()` | 将模型设置为**评估/推理模式** (`training=False`)。 | **测试或验证前必须调用**，防止 Dropout 扔掉数据。 |
| **设备与类型** | `to(device)` | 将模型参数和 buffers **移动到指定设备** (如 GPU/CPU) 或转换类型。 | 常用：`model.to('cuda:0')` 或 `model.half()` (转半精度)。 |
| **保存与加载** | `state_dict()` | **导出权重字典**：返回包含所有 `parameters` 和 `buffers` 的字典。 | **保存模型最推荐的方法**：`torch.save(model.state_dict(), 'xxx.pth')`。 |
| **保存与加载** | `load_state_dict(dict)` | **导入权重字典**：把保存的字典数据重新加载到模型实例中。 | 模型结构必须与保存时完全一致。 |
| **梯度管理** | `zero_grad()` | **清空历史梯度**：把所有参数的 `.grad` 属性清零或设为 None。 | **反向传播 (`loss.backward()`) 前必须调用**，否则梯度会累加。 |

**💡 笔记小贴士：**
你可以把 `parameters` 想象成你的**存款**（天天盯着，随时变化），`buffers` 想象成你的**房产证**（不天天变，但很重要必须保存），而 `modules` 就是装这些东西的**钱包和保险柜**。

````python
#5 梯度与计算控制

tensor1 = torch.rand(2,3, requires_grad=True) #创建一个需要梯度的张量

model = torch.nn.Sequential(
    torch.nn.Linear(3,4),
    torch.nn.BatchNorm1d(4),
    torch.nn.ReLU(),
)

citeration = torch.nn.CrossEntropyLoss()

optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

labels = torch.tensor([0,1]) #假设标签

for param in model.buffers():
    print(param)

print(model.modules)#查看模型的参数和缓冲区，参数是可训练的，缓冲区是不可训练的

dict = model.state_dict()#导出权重字典，包含模型的参数和缓冲区，参数是可训练的，缓冲区是不可训练的
print(dict)
````

````python
# 训练以及预测

epochs = 10
for epoch in range(epochs):
    optimizer.zero_grad()
    y_hat = model(tensor1)
    loss = citeration(y_hat,labels) #假设标签
    loss.backward() #反向传播计算梯度，然后释放计算图
    optimizer.step() #更新模型参数，梯度下降

with torch.no_grad(): #在这个上下文管理器中，所有的计算都不会被记录到计算图中，也就是说不会计算梯度，适用于推理阶段或者不需要梯度的计算
    y_hat = model(tensor1)
    print(y_hat)
````

**广播** 是 PyTorch、NumPy 等库中的一种机制，允许**不同形状的张量进行逐元素运算**。它会自动将较小的张量“扩展”成与较大张量相同的形状，而不需要显式复制数据。

### 核心规则（从尾部维度开始匹配）

1. **从最后一个维度开始往前比较**。
2. 两个维度满足以下条件之一即可兼容：
   - 相等；
   - 其中一个为 `1`；
   - 其中一个维度不存在（视为 `1`）。
3. 不满足上述条件 → 报错 `RuntimeError: The size of tensor a ... must match ...`

# 🧠 PyTorch 计算图速记

## 🔷 1. 计算图的基本结构

* **节点（Node）**：表示数据（Tensor）和计算操作（Op）
* **边（Edge）**：表示数据的流动方向

---

## 🔷 2. 叶子节点与中间节点

* **叶子节点（Leaf）**

  * 用户创建（如 `requires_grad=True`）
  * **梯度会保留在 `.grad` 中**

* **非叶子节点（中间节点）**

  * 由运算产生
  * **默认不保留梯度**
  * 如需保留：使用 `retain_grad()`

---

## 🔷 3. grad_fn

* 表示：**该非叶子节点是通过什么运算得到的**
* 本质：计算图中的一个节点（记录反向传播信息）
* 特点：

  * 叶子节点：`grad_fn = None`
  * 非叶子节点：有 `grad_fn`
* 作用：在反向传播中用于**链式求导**

---

## 🔷 4. 计算图什么时候创建？什么是动态图？

* **创建时机**：forward 过程中边计算边创建
* **动态图**：

  * 每次运行（每个 batch）都会重新构建一张图
  * 运行完即销毁（默认）

---

## 🔷 5. backward 后图和梯度状态

执行：

```python
loss.backward()
```

* **计算图**：默认释放（节省内存）
* **梯度（grad）**：只保留在叶子节点
* **grad_fn**：仍存在，但图结构已不可复用
* 如需保留计算图：`retain_graph=True`

---

## 🔷 6. torch.no_grad()

```python
with torch.no_grad():
```

* 不构建计算图
* 不记录 `grad_fn`
* 不计算梯度
* 等价于：`requires_grad=False`
* 用途：推理阶段（节省显存和计算）
* 同理，当所有参与运算的叶子结点的Tensor 都是 **requires_grad=False**时，也不会创建计算图。

---

## 🔷 7. 前向传播是否需要计算图？

* ❌ 不需要（如果不做反向传播）
* ✔️ 计算图只用于梯度计算
* 👉 推理时应关闭计算图（`no_grad`）

---

# 🔥 总结

👉 **训练：forward 建图 + backward 用图
👉 推理：只 forward，不建图**


````python
w = torch.tensor([1.], requires_grad=True)
x = torch.tensor([2.], requires_grad=True)

a = torch.add(w, x)
# a.retain_grad() #保留中间变量的梯度，默认情况下，只有叶子结点的梯度会被保留，如果需要保留中间变量的梯度，需要调用retain_grad()方法
b = torch.add(w, 1)     # retain_grad()
y = torch.mul(a, b)

y.backward()
print(w.grad)

# 查看叶子结点
print("is_leaf:\n", w.is_leaf, x.is_leaf, a.is_leaf, b.is_leaf, y.is_leaf)
# 查看梯度
print("gradient:\n", w.grad, x.grad, a.grad, b.grad, y.grad)
# 查看 grad_fn
print("grad_fn:\n", w.grad_fn, x.grad_fn, a.grad_fn, b.grad_fn, y.grad_fn)
````
