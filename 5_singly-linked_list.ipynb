1.5.1 无序表实现

```python
class Unorder:
    def __init__(self):
        self.head = None

    def add(self, data):
        temp = Node(data)
        temp.next = self.head
        self.head = temp

    def size(self):
        current = self.head
        count = 0
        while current:
            count += 1
            current = current.next
        return count

    def remove(self, data):#删除节点后修改current即可，previous不要修改
        current = self.head
        previous = None
        while current:
            if current.data == data:
                if current == self.head:
                   self.head = current.next
                   current = current.next
                else:
                    previous.next = current.next
                    current = current.next
            else:
                previous = current
                current = current.next

    def search(self, data):
        current = self.head
        find_list = []
        count = 0
        while current:
            if current.data == data:
                find_list.append(count)
            current = current.next
            count += 1

        if len(find_list) == 0:
            return False
        else:
            return find_list


class Node:
    def __init__(self, data=0, next=None):#节点可能被赋值为0，但并不代表着节点为空
        self.data = data
        self.next = next
```

1.5.2 有序表

```python
class orderList:
    def __init__(self):
        self.head = None
    def isEmpty(self):
        return self.head == None

    def add(self, data):
        current =self.head
        previous=None
        found = False
        while current and not found :
            if current.data <= data:
                previous = current
                current = current.next
            else:
                found = True

        temp = Node(data)
        if previous == None:
            temp.next = self.head
            self.head = temp
        else:
            temp.next = previous.next
            previous.next = temp

    def search(self, data):
        current = self.head
        found = False
        stop = False
        while current and not found and not stop:
            if current.data < data:
                current = current.next
            else:
                if current.data >data:
                    stop = True
                else:
                    found = True
        return found

    def remove(self, data):
        current = self.head
        previous = None
        found = False
        stop = False
        while current and not found and not stop:
            if current.data < data:
                previous = current
                current = current.next
            elif current.data > data:
                stop = True
            else :
                found = True

        if found == True:
            if previous == None:
                self.head = current.next
            else:
                previous.next = current.next

        return found


```

对于self.head（头节点）而言，return self.head就可以把整个链表打印出来，打印链表要保证对链表的头结点的引用保持不变。
