```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        return self.items.pop()

    def isEmpty(self):
        return len(self.items) == 0 #空列表的布尔值为False

    def peektop(self):
        return self.items[len(self.items)-1]

    def size(self):
        return len(self.items)
```

```python
def pay_checker(string):
    s = Stack()
    flag = True
    index = 0

    while index < len(string) and flag:
        if string[index] in "([{":
            s.push(string[index])
        else:
            if s.isEmpty():
                flag = False
            else:
                symbol=s.pop()
                if not match(symbol, string[index]):
                    flag = False
        index += 1

    if flag==True and s.isEmpty():
       return True
    else:
        return False
```

```python
def match(x,y):
    opener = "([{" #用字符串而非列表
    closer = ")]}"

    return opener.index(x) == closer.index(y)
```

```python
string ="{[(){}]}"
print(pay_checker(string))
```

```python
def Devideby16(number):
    s=Stack()
    l =""
    string1="0123456789ABCDEF"
    while number>0:
        s.push(number%16)
        number=number//16

    while not s.isEmpty():
        l = l + string1[s.pop()]#调用函数用(),取索引值用[].
    return l
```

```python
number = 233
print(Devideby16(number))
```

```python
def exchange(string1):
    dict = {}
    dict["("] = 1
    dict["+"] = 2
    dict["-"] = 2
    dict["*"] = 3
    dict["/"] = 3

    start_list = string1.split()#没有参数时，以空格对字符串进行分割,返回的是一个列表,每个元素为字符格式。
    s = Stack()
    finish_list = []

    for item in start_list:
        if item in "ABCDEFGHIJ" or item in "0123456789":
            finish_list.append(item)
        elif item =="(":
            s.push(item)
        elif item == ")":
            pop = s.pop()
            while pop != "(":
                finish_list.append(pop)
                pop = s.pop()
        else:
            while not s.isEmpty() and dict[s.peektop()] > dict[item]:
                finish_list.append(s.pop())
            s.push(item)
    while not s.isEmpty():
        finish_list.append(s.pop())

    return "".join(finish_list)  #"分隔符".join(列表),将列表中的元素以指定的分隔符拼接在一起,要求元素类型为字符

string = "( A * B ) + C * D"
print(exchange(string))
```

```python
def domath(op,op1,op2):
    if op == "+":
        return op1 + op2
    if op == "-":
        return op1 - op2
    if op == "*":
        return op1 * op2
    if op == "/":
        return op1 / op2

def reexchange(string):
    s = Stack()
    start_list = string.split()
    for item in start_list:
        if item in "+-*/":
            op2 = s.pop()
            op1 = s.pop()
            s.push(str(domath(item,int(op1),int(op2))))
        else :
            s.push(item)

    return s.pop()

string = "4 5 16 * +"
print(reexchange(string))
```
