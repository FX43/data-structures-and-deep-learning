1.python数据类型的性能

```python
def test1():
    l=[]
    for i in range(1000):
        l = l+[i]

def test2():
    l=[]
    for i in range(1000):
        l.append(i)

def test3():
    l=[i for i in range(1000)]


def test4():
    l=list(range(1000))
```

```python
from timeit import Timer
if __name__ == "__main__":
    t1 = Timer("test1()", "from __main__ import test1")
    print("concat %f second\n" % t1.timeit(number=1000))

    t2 = Timer("test2()", "from __main__ import test2")
    print("append %f second\n" % t2.timeit(number=1000))

    t3 = Timer("test3()", "from __main__ import test3")
    print("list comprehension %f second\n" % t3.timeit(number=1000))

    t4 = Timer("test4()", "from __main__ import test4")
    print("list(range()) %f second\n" % t4.timeit(number=1000))

```

from __main__ import test3 的含义为从当前的模块下导入相应的函数，如果是其他模块的话语法实例为“from math import sqrt”，对于Timer函数而言，第一个参数为需要反复运行的语句，第二个参数为只需运行一次的安装语句。

```python
popzero =Timer("x.pop(0)", "from __main__ import x")

pop =Timer("x.pop()", "from __main__ import x")

for i in range(1000000,1000000000001):
    x=list(range(i))
    pt = popzero.timeit(number=1000)
    x =list(range(i))
    pz = pop.timeit(number=1000)
    print("%f %f" % (pt, pz))
```
