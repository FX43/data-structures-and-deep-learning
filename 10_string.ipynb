```python
class string():
    def __init__(self):
        self.string = ""
    def StrAssign(self,char):
        self.string = char
    def Strcopy(self,temp):
        temp = self.string
        return temp
    def StrEmpty(self):
        return not self.string
    def StrCompare(self,temp):
        return len(self.string) >= len(temp)
    def StrLength(self):
        return len(self.string)
    def Substring(self,pos,num):
        return self.string[pos:pos+num]
    def Concat(self,string1,string2):
        self.string = string1 + string2
        return self.string
    def Clearstring(self):
        self.string = ""
    def Destorystring(self):
        self.string = None
```

#1.暴力匹配算法：（时间复杂度为O(m*n)）

当匹配失败时，此时i指向主串中匹配失败的字符，j代表着虽然这次匹配失败了，但之前已经成功匹配了j-1次，这次比较是从主串的i-(j-1)开始的，因此下次匹配要从i-j+2位置开始

j既代表着当前待比较字符的索引，也在一定程度上给出了已经成功匹配的字符串长度

        int Index(SString S, SString T){
            int i = 1, j = 1;
            while(i <= S.length && j <= T.length){
                if(S.ch[i] == T.ch[j]){
                    ++i; ++j;	//继续比较后继字符
                }else{
                    //指针后退重新开始匹配
                    i = i-j+2;
                    j = 1;
                }
            }
            if(j > T.length){代表着匹配成功，此时i指向的字符不包括在匹配成功的字符里
                return i - T.length;
            }else{
                return 0;
            }
        }


```python
def BF(string,substring):
    length1 = len(string)
    length2 = len(substring)
    i,j = 0,0
    while i < length1 and j < length2:
        if string[i] == substring[j]:
            i+=1
            j+=1
        else:
            i = i - j + 1
            j = 0
    if j >= length2:
        return i - length2
    else:
        return False

string = "ABACABAD"
substring = "BA"
print(BF(string,substring))
```

#2.kmp算法：

主串指针不后退，只有子串指针会后退

```python
def nextval(substring):
    next = [0]
    prefix_length = 0
    length = len(substring)
    i = 1
    while i < length:
        if substring[i] == substring[prefix_length]:
            prefix_length += 1
            next.append(prefix_length)
            i = i + 1
        else:
            if not prefix_length:
                next.append(prefix_length)
                i = i + 1
            else:
                prefix_length = next[prefix_length-1]
    return next

def kmp(string,substring):
    next_val = nextval(substring)
    length1 = len(string)
    length2 = len(substring)
    i,j = 0,0
    while i < length1 and j < length2:
        if string[i] == substring[j]:
            i+=1
            j+=1
        else:
            if j >0:#字串跳过匹配的个数，例如2就代表着跳过两个字符，从第三个字符开始匹配（索引为2）
                j = next_val[j-1]
            else:
                i = i + 1#j等于0说明第一个字符就无法匹配，直接让主串指针前进

    if j >= length2:
        return i - length2
    else:
        return False

string = "ABACABAD"
substring = "BA"
print(kmp(string,substring))
# print(nextval(substring))
```
