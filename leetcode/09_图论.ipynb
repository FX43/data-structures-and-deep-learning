## 200-1 岛屿数量


### 解题思路


### 方法一：DFS


````python
class Solution(object):
    def numIslands(self, grid):
        """
        :type grid: List[List[str]]
        :rtype: int
        """
        m = len(grid)
        n = len(grid[0])
        num = 0

        def dfs(i,j):
            if i < 0 or i > m-1 or j < 0 or j > n-1:
                return
            if grid[i][j] == "0" or grid[i][j] == "2":
                return
            grid[i][j] = "2"
            dfs(i-1,j)
            dfs(i+1,j)
            dfs(i,j+1)
            dfs(i,j-1)

        for i in range(m):
            for j in range(n):
                if grid[i][j] == "1":
                    #下面不能写grid[i][j] = "0"，否则进入递归后直接退出
                    dfs(i,j)
                    num += 1

        return num
````

### 方法二：BFS  对于坐标而言，用元组存储更加方便

deque（双端队列数据结构）的用法：

基本操作：

        quene = deque([i,j]): 初始化
        d.appendleft() :从左端添加元素
        d.append():从右端添加元素
        d.pop(): 从右端弹出元素
        d.popleft(): 从左端弹出元素

优势：
    双端队列两端入队与出队的时间复杂度都为O(1)


````python
from collections import deque

class Solution(object):
    def numIslands(self, grid):
        if not grid:
            return 0

        m = len(grid)
        n = len(grid[0])

        def bfs(i, j):
            queue = deque([(i, j)]) #注意这里,不能是deque([i,j]),这样每个元素就是i,j,而不是(i,j)这个元组
            grid[i][j] = "0"  # 标记为已访问

            while queue:
                x, y = queue.popleft()#这种情况下x与y都是数字类型，相当于提前解包，如果左端只有一个变量的话那么它是一个列表或者元组
                # 四个方向
                for dx, dy in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
                    nx, ny = x + dx, y + dy
                    if 0 <= nx < m and 0 <= ny < n and grid[nx][ny] == "1":#坐标要合理,同时是没有被探索过的坐标
                        grid[nx][ny] = "0"
                        queue.append((nx, ny))

        num = 0
        for i in range(m):
            for j in range(n):
                if grid[i][j] == "1":
                    bfs(i, j)
                    num += 1

        return num
````

## 994-2 腐烂的橘子


### 腐烂橘子问题

> **曾错重点：第一次完整扫描网格后，进入 BFS 前，必须先判断新鲜橘子数 `num == 0`，为 0 立即返回 `0`。**
>
> 判断应放在双重 `for` 循环全部结束之后，而不是只扫描完一行时。不能只在 `while queue` 内判断：例如 `[[0]]` 的队列为空，循环不会执行，漏判会误返回 `-1`。

```python
# 完整统计网格之后，BFS 之前
if not num:
    return 0
```


````python
class Solution:
    def orangesRotting(self, grid: List[List[int]]) -> int:
        m,n = len(grid),len(grid[0])
        queue = deque()
        num,minute = 0,0 #用于记录好橘子以及腐烂总计时长
        for i in range(m):
            for j in range(n):
                if grid[i][j] == 2:
                    queue.append([i,j])
                if grid[i][j] == 1:
                    num += 1

        if not num: #在判断完第一个for循环后要先判断是否有新鲜橘子,如果没有的话直接返回0
            return 0

        while queue:
            if not num:
                return minute
            length = len(queue)
            for _ in range(length):
                x,y = queue.popleft()
                for dx,dy in [(0,1),(0,-1),(1,0),(-1,0)]:
                    nx,ny = x+dx,y+dy
                    if 0 <= nx < m and 0<= ny < n and grid[nx][ny] == 1:
                        grid[nx][ny] = 2
                        queue.append([nx,ny])
                        num -= 1
            minute += 1
        return -1
````

## 207-3 课程表


### 课程表问题（拓扑排序问题）

**采用入度出度算法**

当从队列中弹出某个元素的时候，只需对这个元素的子节点的入度减一即可，使用邻接表数据结构保存每个节点的子节点都有谁，在进行入度减一操作时直接在邻接表中进行而不需要遍历所有节点

对于邻接表数据结构：索引代表某个节点，索引上的列表中是这个节点所有的子节点


````python
class Solution(object):
    def canFinish(self, numCourses, prerequisites):
        """
        :type numCourses: int
        :type prerequisites: List[List[int]]
        :rtype: bool
        """
        indegree = [0] * numCourses
        adject = [[] for _ in range(numCourses)]
        num = 0

        for child,par in prerequisites:
            adject[par].append(child)#存储这个节点的子节点，方便后续子节点的入度减一操作
            indegree[child] += 1

        queue = deque([i for i in range(numCourses) if indegree[i]==0])#构建一个双端队列，其中里面的元素都入读都为0

        while queue:
            pop = queue.popleft()
            num += 1 #+1操作在这里而不是在下面,如果在加入队列的时候加1,那么最开始入度为0的元素没有被统计到

            for j in adject[pop]:
                indegree[j] -= 1
                if indegree[j] == 0:
                    queue.append(j)

        return True if num == numCourses else False
````

## 208-4 实现 Trie (前缀树)


### 生成前缀树

生成前缀树的思想核心是trienode节点，这个节点本身代表一个字母，**哈希表里面存储的是以这个字母为前缀的其他字母(也就是trienode节点)**，通过哈希表可以直接访问到这些字母节点




````python
class Trienode:
    def __init__(self):
        self.hashtable = {}#对于这个结构来说,键为字母前缀,值为对应的Trienode对象
        self.flag = False

class Trie:

    def __init__(self):
        self.root = Trienode() #根节点的作用在于存储每个单词的第一个字母

    def insert(self, word: str) -> None:
        node = self.root
        for char in word:
            if char not in node.hashtable:
                node.hashtable[char] = Trienode()
            node = node.hashtable[char]
        node.flag = True

    def search(self, word: str) -> bool:
        node = self.root
        for char in word:
            if char not in node.hashtable:
                return False
            node = node.hashtable[char]
        return True if node.flag else False

    def startsWith(self, prefix: str) -> bool:
        node = self.root
        for char in prefix:
            if char not in node.hashtable:
                return False
            node = node.hashtable[char]
        return True
````
