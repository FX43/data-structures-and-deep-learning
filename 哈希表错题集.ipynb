_**#1.最长连续序列**_

```python
class Solution(object):
    def longestConsecutive(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        hashtable = set(nums)
        max_sum = 0
        for i in hashtable:#后续的i+=1不会影响到这里
            if i-1 not in hashtable:#如果一个数的前面不存在的话，那么把它作为起点
                curr_sum = 1
                while i+1 in hashtable:
                    curr_sum += 1
                    i += 1
                max_sum = max(max_sum,curr_sum)#当前最大值与历史最大值比较
        return max_sum
```
